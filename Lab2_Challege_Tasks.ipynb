```python
# [EXERCISE 1] REDACTION PROTOCOL
target_name = "ELIAS"   
SEED_NUM = 0
redacted_name = ""

for ch in target_name:
    if ch in "AEIOU":        # Check if vowel
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += ch

print("Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Redacted Output:", redacted_name)
```

```python
# [EXERCISE 2] VERIFICATION HASH
engineer_name = "ELIAS"   
SEED_NUM = 0             
loops = SEED_NUM + 3             

# Step 1: Get ASCII of first letter
first_letter = engineer_name[0]
ascii_val = ord(first_letter)

# Step 2: Compute verification hash
verification_hash = (ascii_val * SEED_NUM) - loops

print("First Letter ASCII:", ascii_val)
print("SEED_NUM:", SEED_NUM)
print("Total Loops:", loops)
print("Verification Hash:", verification_hash)
```

```python
# [EXERCISE 3] SECURE INPUT LOOP
correct_password = "ELIAS321"  
attempts = 0

while True:
    user_input = input("Enter password: ")
    attempts += 1
    
    if user_input == correct_password:
        print("Access Granted!")
        break
    else:
        print("Incorrect password, try again...")

print("Correct Password Set:", correct_password)
print("Input Attempts:", attempts)
print("Execution Log: Loop repeated until correct password entered")
print("Attempts Before Success:", attempts)
```
